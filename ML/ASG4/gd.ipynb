```python
def function(x):
    return (x+3)**2

def derivative(x):
    return 2*(x+3)
```

```python
import matplotlib.pyplot as plt
```

```python
x=2
learning_rate=0.2
iterations=50
```

```python
print("Gradient Descent")
```

```python
x_vals=[]
y_vals=[]

for i in range(iterations):
    gradient=derivative(x)
    x=x-learning_rate*gradient
    y=function(x)
    x_vals.append(x)
    y_vals.append(y)
    print(f"Iteration {i+1} : x = {x}   y = {y }")
```

```python
plt.plot(x_vals,y_vals,marker='o')
plt.show()
```
