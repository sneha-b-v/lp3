```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score,f1_score,recall_score,precision_score,confusion_matrix)
from nltk.corpus import stopwords
```

```python
file=r"C:\Users\Sneha\Desktop\BE\sem1\lp3\emails.csv"
df=pd.read_csv(file)
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.dropna(inplace=True)
```

```python
df.drop('Email No.',axis=1,inplace=True)
```

```python
stopwords_arr=set(stopwords.words("english"))

cols_to_drop=[
    col for col in df.columns if col in stopwords_arr or len(col)<=2
]

df.drop(cols_to_drop,axis=1,inplace=True)
```

```python
sc=StandardScaler()

y=df["Prediction"]
x=df.drop("Prediction", axis=1)
x=x.apply(pd.to_numeric,errors="coerce")

x_train, x_test, y_train, y_test= train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)

x_train_sc=sc.fit_transform(x_train)
x_test_sc=sc.transform(x_test)

model=KNeighborsClassifier(n_neighbors=6)
model2=SVC()
model.fit(x_train_sc,y_train)

k_pred=model.predict(x_test_sc)


cf=confusion_matrix(y_test,k_pred)
ac=accuracy_score(y_test, k_pred)

model2.fit(x_train_sc,y_train)

s_pred=model2.predict(x_test_sc)


cf2=confusion_matrix(y_test,s_pred)
ac2=accuracy_score(y_test,s_pred)
```

```python
print(cf)
print("accuracy")
print(ac)
print("===========")
print(cf2)
print("accuracy_score")
print(ac2)
```
